```python
!pip install ultralytics -q

from ultralytics import YOLO
from google.colab import files
import glob

model = YOLO('yolov8n.pt')

videos = glob.glob('/content/*.mov') + glob.glob('/content/*.mp4')
if videos:
    video_path = videos[0]
    print("Processing video:", video_path)

    results = model.track(source=video_path, save=True, tracker="bytetrack.yaml", classes=[2, 7])

    output_videos = glob.glob('/content/runs/detect/track*/*.mp4')
    if output_videos:
        print("Download started for the final video.")
        files.download(output_videos[-1])
else:
    print("No video file found in the directory.")
```
